```python
from pyspark.sql.functions import when

df = spark.table("silver_customers_1")

df = df.withColumn(
    "age_group",
    when(df.age < 30, "Young")
    .when(df.age < 40, "Adult")
    .otherwise("Senior")
)

df.write.mode("overwrite").saveAsTable("silver_customers_2")

print("Silver-2 completed")
```
